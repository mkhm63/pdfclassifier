```python
import os
import pandas as pd
import random

# Define folder where your .txt files are
text_dir = "output/texts"
output_file = "output/labels.csv"

# Create output folder if it doesn't exist
os.makedirs("output", exist_ok=True)

# List all .txt files in the folder
filenames = [f for f in os.listdir(text_dir) if f.endswith(".txt")]

# Generate random labels for testing
sample_labels = ["high", "medium", "low"]
labels = [random.choice(sample_labels) for _ in filenames]

# Create DataFrame and save to CSV
df = pd.DataFrame({"filename": filenames, "label": labels})
df.to_csv(output_file, index=False)

# Show first few rows
df.head()
```

```python
import os
import pandas as pd

# Load existing labels if any
existing_labels = pd.read_csv("output/labels.csv") if os.path.exists("output/labels.csv") else pd.DataFrame()

# Find new .txt files not yet labeled
existing_files = set(existing_labels["filename"]) if not existing_labels.empty else set()
all_files = set(f for f in os.listdir("output/texts") if f.endswith(".txt"))
new_files = sorted(all_files - existing_files)

# Create a new DataFrame with blank labels for new files
new_df = pd.DataFrame({"filename": list(new_files), "label": [""] * len(new_files)})

# Merge with existing labels
combined_df = pd.concat([existing_labels, new_df], ignore_index=True)
combined_df.to_csv("output/labels.csv", index=False)
combined_df.tail()
```

```python
from wordcloud import WordCloud, STOPWORDS

def plot_wordclouds_by_label(label_texts):
    stopwords = set(STOPWORDS)
    stopwords.update(["et", "al", "data", "using", "study", "based", "used", "covid", "results", "conclusion"])

    for label, texts in label_texts.items():
        if texts:
            full_text = " ".join(texts).lower()
            wc = WordCloud(
                width=800,
                height=400,
                background_color="white",
                stopwords=stopwords
            ).generate(full_text)

            plt.figure(figsize=(10, 5))
            plt.imshow(wc, interpolation="bilinear")
            plt.axis("off")
            plt.title(f"Word Cloud: {label}")
            plt.tight_layout()
            plt.show()
```
